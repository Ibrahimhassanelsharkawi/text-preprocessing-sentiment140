# Task: N-Gram Language Model
Calculate the probability of 3 sentences from the dataset using the Markov assumption.

**Method:** Bigram Model and Maximum Likelihood Estimation (MLE).


```python
import pandas as pd
from collections import Counter

df = pd.read_csv("Data/preprocessed.csv")
texts = df["clean_text"].dropna().astype(str)
print("Total rows:", len(df))
```

```python
unigram_counts = Counter()
bigram_counts = Counter()

for text in texts:
    words = text.split()
    if not words:
        continue
    sentence = ["<s>"] + words + ["</s>"]
    unigram_counts.update(sentence[:-1])
    bigram_counts.update(zip(sentence[:-1], sentence[1:]))
```

## 3. Markov Assumption and MLE

Under the Bigram Markov Assumption:

$$P(w_i\mid w_{i-1})=\frac{C(w_{i-1},w_i)}{C(w_{i-1})}$$



```python
sentences = [
    "happy mother day",
    "want go home",
    "think getting sick"
]

for i, text in enumerate(sentences, 1):
    words = ["<s>"] + text.split() + ["</s>"]
    probability = 1.0
    print(f"\nSentence {i}: {text}")
    print("Sequence:", " ".join(words))
    print("Calculation steps:")
    for previous, current in zip(words[:-1], words[1:]):
        numerator = bigram_counts[(previous, current)]
        denominator = unigram_counts[previous]
        conditional_probability = numerator / denominator if denominator else 0
        probability *= conditional_probability
        print(f"P({current} | {previous}) = {numerator}/{denominator} = {conditional_probability:.10f}")
    print(f"Final sentence probability = {probability:.12g}")
```
